# Drivers: horas y millas por conductor

Se cargan los archivos `drivers.csv` y `timesheet.csv`, se suman las horas y millas registradas por cada conductor y se combinan con sus datos.

```python
import pandas as pd

drivers = pd.read_csv("../data/drivers.csv")
timesheet = pd.read_csv("../data/timesheet.csv")
drivers.head()
```

```python
timesheet.head()
```

```python
summary = (
    timesheet.groupby("driverId", as_index=False)[["hours-logged", "miles-logged"]]
    .sum()
)
summary.head()
```

```python
result = pd.merge(
    drivers[["driverId", "name"]],
    summary,
    on="driverId",
)
result.sort_values("miles-logged", ascending=False).head(10)
```
